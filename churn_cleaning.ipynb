```python
import pandas as pd

df = pd.read_csv("Telco-Customer-Churn.csv")

# 1. Fix TotalCharges: convert to numbers; blanks become NaN, then 0
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce").fillna(0)

# 2. Convert SeniorCitizen from 0/1 to No/Yes (easier to read in charts)
df["SeniorCitizen"] = df["SeniorCitizen"].map({0: "No", 1: "Yes"})

# 3. Create a tenure group column (useful for SQL, Excel, and Power BI)
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0-12 months", "13-24 months", "25-48 months", "49-72 months"]
)

# 4. Add a numeric churn flag (makes churn rate easy to calculate)
df["ChurnFlag"] = (df["Churn"] == "Yes").astype(int)

# 5. Check and save
print(df.dtypes)
print(df.isnull().sum().sum(), "nulls")
df.to_csv("churn_clean.csv", index=False)
```
